#### Performing ROW_NUMBER, RANK, DENSE_RANK operations using Window function

```
from pyspark.sql.window import Window

window = Window.partitionBy("DepartmentId").orderBy("Salary")
emp_df.withColumn('ROW NUMBER', row_number().over(window))\
    .withColumn('RANK', rank().over(window))\
    .withColumn('DENSE RANK', dense_rank().over(window))\
    .display()
```

```
# Find top 2 salary earner from each department

from pyspark.sql.window import Window

window = Window.partitionBy("DepartmentId").orderBy(desc("Salary"))
emp_df.withColumn('DENSE RANK', dense_rank().over(window))\
    .filter(col('DENSE RANK')<=2)\
    .display()
```
