```
# Finding Unique Records
emp_df.distinct().show(100)
```

```
emp_df.distinct().count()
emp_df.count()
```

```
# Dropping ducpliates

emp_df.drop_duplicates(["id", "name", "age", "salary", "country", "department"]).show(100)
```

```
# Sorting the data on the basis of salary
emp_df.sort("salary").show(100)
```

```
# Using col to sort in descending order, it is usually not possible to sort using only contraints
emp_df.sort(col("salary").desc()).show()
```
