```
/FileStore/tables/userdata.parquet
```

```
df = spark.read.parquet("/FileStore/tables/userdata.parquet")
df.show(5)
```

```
# Writing to Parquet Format
df.write.format("parquet")\
        .option("header","true")\
        .option("mode","overwrite")\
        .option("path","/FileStore/tables/userdata_parquet")\
        .save()
```

```
# Way to view the written file
dbutils.fs.ls("/FileStore/tables/userdata_parquet")
```

```
# writing with comperssion
df.write.format("parquet")\
        .option("header","true")\
        .option("mode","ignore")\
        .option("compression","snappy")\
        .option("path","/FileStore/tables/compressed_parquet")\
        .save()
```

```
dbutils.fs.ls("/FileStore/tables/compressed_parquet")
```
