```
from pyspark.sql.functions import *
from pyspark.sql.types import *
emp_data = [
    (1, "Alice", 30, 50000, "USA", "Engineering"),
    (2, "Bob", None, 40000, "Canada", "Marketing"),
    (3, "Cathy", 35, 60000, "UK", "HR"),
    (4, "David", 28, 45000, "India", "Engineering"),
    (5, "Eve", 70, 70000, "Germany", "Finance"),
    (6, "Frank", 32, 48000, "Australia", "Marketing"),
    (7, "Grace", 79, 52000, "France", "Engineering"),
    (8, "Hank", 83, 62000, "USA", "Finance"),
    (9, "Ivy", 27, 55000, "India", "HR"),
    (10, "Jack", 36, 58000, "Canada", "Engineering")
]
emp_schema = ["id", "name", "age", "salary", "country", "department"]
emp_df = spark.createDataFrame(data=emp_data, schema=emp_schema)
emp_df.createOrReplaceTempView("emp_df_temp_view")
emp_df.show()
```

```
# Use of when-otherwise in sparks
emp_df.withColumn("adult",when(col("age")>=30 , "Yes")
                        .when(col("age")<30, "No")
                        .otherwise("Novalue")).show()
```

```
# Now condition is to fix the null value then make a new column
emp_df.withColumn("age",when(col("age").isNull(), lit(30))
                        .otherwise(col("age")))\
        .withColumn("adult",when(col("age")>=30, "Yes")
                            .otherwise("No")).show()
```

```
# Now I have a condition where i need to make a column of category of adult and senior citizens
emp_df.withColumn("category",when((col('age')>30) & (col('age')<80) , "Sr. Citizen")
                            .when(col('age')<=30, "adult")
                            .otherwise('sr.sr.citizen')).show()
```
