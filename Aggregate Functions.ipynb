```
from pyspark.sql.functions import *
from pyspark.sql.types import *
emp_data = [
    (1, "Alice", 30, 50000, "USA", "Engineering"),
    (2, "Bob", None, 40000, "Canada", "Marketing"),
    (3, "Cathy", 35, 60000, "UK", "HR"),
    (4, "David", 28, 45000, "India", "Engineering"),
    (5, "Eve", 70, 70000, "Germany", "Finance"),
    (6, "Frank", 32, 48000, "Australia", "Marketing"),
    (7, "Grace", 79, 52000, "France", "Engineering"),
    (8, "Hank", 83, 62000, "USA", "Finance"),
    (None, None, None, None, None, None),
    (9, "Ivy", 27, 55000, "India", "HR"),
    (10, "Jack", 36, 58000, "Canada", "Engineering")
]
emp_schema = ["id", "name", "age", "salary", "country", "department"]
emp_df = spark.createDataFrame(data=emp_data, schema=emp_schema)
emp_df.createOrReplaceTempView("emp_df_temp_view")
emp_df.show()
```

```
# It counts all the records including Null
emp_df.count()
```

```
# It skips null value in name column and shows the count
emp_df.select(count(col('name'))).show()
```

```
# Using sum, max, min operations
emp_df.select(sum('salary').alias("Total Salary"), max('salary').alias('Maximum Salary'), min('salary').alias('Minimum Salary')).show()
```

```
# Calculating the average and casting the floating value into the integer
emp_df.select(sum('salary'), count('salary'), avg('salary').cast('int')).show()
```
