```
youtube_df = spark.read.format("csv")\
            .option("header","true")\
            .option("inferschema","false")\
            .option("mode","FAILFAST")\
            .load("/FileStore/tables/youtube_subscribers_data.csv")
youtube_df.show(100)
            
```

```
# Now my goal is to do partition based on the primary language on above dataset
youtube_df.write.format("csv")\
                .option("header","true")\
                .option("mode","overwrite")\
                .option("path","/FileStore/tables/partionBY_language_data.csv")\
                .partitionBy("Primary language")\
                .save()
```

```
dbutils.fs.ls("/FileStore/tables/partionBY_language_data.csv")
# Here we are seeing partition are made based on the primary language from our youtube_subscriber datasets
```

```
# I want to do the partitioning based on the country and primary language

youtube_df.write.format("csv")\
                .option("header","true")\
                .option("mode","overwrite")\
                .option("path","/FileStore/tables/partitionBY_country_language_data.csv")\
                .partitionBy("country","Primary language")\
                .save()
```

```
dbutils.fs.ls("/FileStore/tables/partitionBY_country_language_data.csv")
```

```
# Inside this path, it made another partition based on the language
dbutils.fs.ls("/FileStore/tables/partitionBY_country_language_data.csv/country=Argentina/")
```

```
# Not I am trying to do bucketing on the above datasets based on subscribers into 10 buckets
# BucketBy only supports saveAsTable instead of save
youtube_df.write.format("csv")\
                .option("header","true")\
                .option("mode","overwrite")\
                .option("path","/FileStore/tables/bucketBy_subscriber_data.csv")\
                .bucketBy(10,"Subscribers (millions)")\
                .saveAsTable("bucketBy_partition_over_subscriber")
```

```
dbutils.fs.ls("/FileStore/tables/bucketBy_subscriber_data.csv")
```
